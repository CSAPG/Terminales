```python
#####################################################
###               Donnees a modifier              ###
### Attention, le separateur decimal est le point ###
#####################################################
### Ici : satellite DSX
## t en min
t=[0,5,10,15,20,25,30]
## latitude en degre
lat=[25.33,24.97,24.59,24.16,23.69,23.18,22.62]
## longitude en degre
lon=[115.02,115.94,116.88,117.84,118.83,119.86,120.93]
## altitude en km
h=[17170.03,17025.56,16844.35,16625.26,16369.31,16076.28,15743.93]
```

```python
### Ne pas modifier
## Rayon terrestre en km
Rt=6378.0
## Jour sideral en min
Js=1436.066
## Vitesse angulaire en degre/min
w=360/Js
## nombre de positions
n=len(t)

## initialisations
r=np.zeros(n)
loncr=np.zeros(n)
latr=np.zeros(n)
x=np.zeros(n)
y=np.zeros(n)
z=np.zeros(n)
d=np.zeros(n-1)
v=np.zeros(n-1)
S=np.zeros(n-1)
s=np.zeros(n-1)
tt=np.zeros(n-1)

## positions dans referentiel geocentrique en km
for i in range(n):
## rayon du satellite
	r[i]=h[i]+Rt
## longitude corrigee de la rotation propre, en rad
	loncr[i]=(lon[i]+w*t[i])/180*pi
## latitude en rad
	latr[i]=lat[i]/180*pi
## coordonnees en km
	x[i]=r[i]*cos(latr[i])*cos(loncr[i])
	y[i]=r[i]*cos(latr[i])*sin(loncr[i])
	z[i]=r[i]*sin(latr[i])

## mouvement
for i in range(n-1):
## Temps sans derniere mesure
	tt[i]=t[i]

### Fonction de calcul de l'aire 
### avec formule de Heron
def aire(a,b,c):
	p=(a+b+c)/2
	S=(p*(p-a)*(p-b)*(p-c))**0.5
	return S
```

```python
####################
#### A COMPLETER ###
####################
for i in range(n-1):
## Distance parcourue entre i et i+1, en km
	d[i]=((x[i+1]-x[i])**2+(y[i+1]-y[i])**2+(z[i+1]-z[i])**2)**0.5
## vitesse en km/min
	v[i]=d[i]/5
## aire balayee en km2
	S[i]=aire(r[i],r[i+1],d[i])
## vitesse areolaire en km2/min
	s[i]=S[i]/5
```

```python
### Ne pas modifier

tmax=max(tt)+2
rmax=max(r)*1.1
vmax=max(v)*1.1
Smax=max(S)*1.1
smax=max(s)*1.1

fig=plt.figure(1,figsize=(4,9))
fig.subplots_adjust(bottom=0.085, left=0.225, top = 0.975, right=0.975)
fig.add_subplot(3,1,1)
plt.plot(t,r,"+")
##plt.title("Rayon en fonction du temps")
plt.xlabel("temps en min")
plt.ylabel("rayon en km")
plt.xlim(0,tmax)
plt.ylim(0,rmax)
plt.grid(True)
fig.add_subplot(3,1,2)
plt.plot(tt,v,"+")
##plt.title("Vitesse en fonction du temps")
plt.xlabel("temps en min")
plt.ylabel("vitesse en km/min")
plt.xlim(0,tmax)
plt.ylim(0,vmax)
plt.grid(True)
fig.add_subplot(3,1,3)
plt.plot(tt,s,"+")
##plt.title("Vitesse areolaire en fonction du temps")
plt.xlabel("temps en min")
plt.ylabel("vitesse areolaire en km^2/min")
plt.xlim(0,tmax)
plt.ylim(0,smax)
plt.grid(True)

plt.show()
```
